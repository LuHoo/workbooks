## Exercise 1.1 Hypergeometric distribution


The Python calculation uses `scipy.stats.hypergeom.pmf()` with four arguments:
$N$, $M$, $n$, and $k$. The `pmf` suffix indicates that the function returns
probabilities for exact outcomes.


```python
hypergeom.pmf(0, M=(17) + (314), n=17, N=60)
hypergeom.pmf(1, M=(17) + (314), n=17, N=60)
```

*Cumulative probabilities* $P(k \leq k)$ (the probability of finding not more
than $k$ errors) are obtained with `scipy.stats.hypergeom.cdf()`. For example,
if we are interested in the probability of finding no more than one error, we
evaluate the cumulative distribution function at $k = 1$.


```python
hypergeom.cdf(1, M=(17) + (314), n=17, N=60)
hypergeom.pmf(0, M=(17) + (314), n=17, N=60) + hypergeom.pmf(1, M=(17) + (314), n=17, N=60)
```

Probabilities in the right-hand tail $P(k > k)$ can be calculated by using the
survival function. For example, the probability $P(k > 1)$ is calculated by


```python
hypergeom.sf(1, M=(17) + (314), n=17, N=60)
```

## Exercise 1.2 Binomial distribution


The relevant Python functions are `scipy.stats.binom.pmf()` for probabilities
$P(k = k)$ and `scipy.stats.binom.cdf()` for cumulative probabilities
$P(k \leq k)$. Both functions use the parameters `k`, `n`, and
$\pi = M / N$.


```python
binom.pmf(1, 60, 17 / 331)
binom.cdf(1, 60, 17 / 331)
```

## Exercise 1.3 Poisson distribution


 
The functions `dpois()` and `ppois()` are used for the calculation of probabilities from the Poisson distribution.
They require two parameter values as arguments: the number of errors $k$ and the value of $\mu = n\pi$.
For the *Case: Number of students* remember that $n = 60$ and $\pi = 17 / 331$.
 


```python
poisson.pmf(1, mu=60 * 17 / 331)
poisson.pmf(3, mu=60 * 17 / 331)
```

 
The probability of finding one error is 14.14%.
 
To calculate the probability of finding no more than one error in the sample of 60 Ph.D.s, use
 


```python
poisson.cdf(1, mu=60 * 17 / 331)
```

## Exercise 1.4 Normal distribution


In Python we use `scipy.stats.norm.cdf()`. The argument `x` is the sampling
result that we evaluate, `loc` is the population mean, and `scale` is the
standard deviation of the mean.


```python
norm.cdf(x= 1012, loc= 1030, scale= 115.26 / sqrt(200))
```

Note that the syntax is more elaborate than what we have done so far. In
Python we can choose between passing argument values by position and passing
values specifically assigned to named arguments.
There are two advantages to using argument names and values. First, the command
is easier to interpret by a reviewer. Second, we can pass the arguments in any
order. The following commands are therefore equivalent:


```python
norm.cdf(1012, 1030, 115.26 / sqrt(200))
norm.cdf(x= 1012, loc= 1030, scale= 115.26 / sqrt(200))
norm.cdf(scale= 115.26 / sqrt(200), loc= 1030, x= 1012)
```

## Exercise 1.5 Student's $t$ distribution


To calculate the probability found of 14.7%, we use
`scipy.stats.t.cdf()` with arguments `x` for the boundary value and `df` for
the number of degrees of freedom.


```python
t_val = (1004 - 1030) / (73.8 / sqrt(10))
t.cdf(t_val, df=9)
```

 
The probability is 14.7%. Notice that we used another shortcut notation: we added brackets to the assignment of the `t_val` variable, so that the assigned value is returned immediately.
 


## Exercise 1.6 $\chi^2$ (chi-squared) distribution


 
The 95\% upper bound on a $\chi^2$ distributed variable with (10 - 1) degrees of freedom is
 


```python
chi2.ppf(.95, df=9)
```

 
Similarly, to calculate the upper bound for $\sigma^2$, we start with the lower bound of a $\chi^2$ distributed variable
 


```python
chi2.ppf(.05, df=9)
```

 
These two results are then used to calculate the upper and lower 90\% confidence bounds of the population variance.
 


## Exercise 1.7 $F$ distribution


Hypothesis testing is done with `scipy.stats.f.ppf()` or
`scipy.stats.f.isf()`. Here we use the upper-tail form because the rejection
region is in the right-hand tail of the $F$ distribution.


```python
f_crit = f.isf(0.05, dfn=25, dfd=23)
```

 
In the example above, the critical value of the $F$ statistic is the value shown in the output. Depending on whether the value of the test statistic is greater or smaller than this value, we then reject the null hypothesis or we do not.
 
The probability of obtaining this particular value of the test statistic is calculated with the `pf` function.
 


```python
f.sf(f_crit, dfn=25, dfd=23)
```
