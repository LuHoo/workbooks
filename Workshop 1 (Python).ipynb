## Exercise 1.1 Hypergeometric distribution


This exercise uses the *Case: Number of students* to calculate probabilities
for the number of errors in a sample without replacement. The practical goal is
to read distribution output as audit evidence: how surprising is a particular
sample result if the population contains 17 errors in 331 students?
The Python calculation uses `scipy.stats.hypergeom.pmf()` with four arguments:
$N$, $M$, $n$, and $k$. The `pmf` suffix indicates that the function returns
probabilities for exact outcomes.


```python
hypergeom.pmf(0, M=(17) + (314), n=17, N=60)
hypergeom.pmf(1, M=(17) + (314), n=17, N=60)
```

*Cumulative probabilities* $P(k \leq k)$ (the probability of finding not more
than $k$ errors) are obtained with `scipy.stats.hypergeom.cdf()`. For example,
if we are interested in the probability of finding no more than one error, we
evaluate the cumulative distribution function at $k = 1$.


```python
ada_set_context("1.1")
hypergeom.cdf(1, M=331, n=17, N=60)
p_zero = hypergeom.pmf(0, M=331, n=17, N=60)
p_one = hypergeom.pmf(1, M=331, n=17, N=60)
p_zero + p_one
```

Probabilities in the right-hand tail $P(k > k)$ can be calculated by using the
survival function. For example, the probability $P(k > 1)$ is calculated by


```python
hypergeom.sf(1, M=(17) + (314), n=17, N=60)
```

The right-tail result is the probability of seeing more than one error. In an audit setting, this tail probability helps students connect a sample finding to the question of whether the observed number of errors is still compatible with the assumed population error count.
 


## Exercise 1.2 Binomial distribution


This exercise repeats the error-count question with the binomial distribution.
The practical task is to see what changes when sampling is treated as
independent trials with a fixed error probability.
The relevant Python functions are `scipy.stats.binom.pmf()` for probabilities
$P(k = k)$ and `scipy.stats.binom.cdf()` for cumulative probabilities
$P(k \leq k)$. Both functions use the parameters `k`, `n`, and
$\pi = M / N$.


```python
binom.pmf(1, 60, 17 / 331)
binom.cdf(1, 60, 17 / 331)
```

Compare the exact and cumulative results with the hypergeometric calculations from Exercise 1.1. The learning point is to notice how the modeling assumption affects the probability attached to the same sample finding.
 


## Exercise 1.3 Poisson distribution


 
This exercise uses the Poisson distribution as another way to approximate rare error counts. The setup stays with the *Case: Number of students*, using the expected number of errors in the sample as $\mu$.
The functions `dpois()` and `ppois()` are used for the calculation of probabilities from the Poisson distribution.
They require two parameter values as arguments: the number of errors $k$ and the value of $\mu = n\pi$.
For the *Case: Number of students* remember that $n = 60$ and $\pi = 17 / 331$.
 


```python
poisson.pmf(1, mu=60 * 17 / 331)
poisson.pmf(3, mu=60 * 17 / 331)
```

 
The probability of finding one error is 14.14%. Read the two outputs as exact-count probabilities before moving to the cumulative calculation.
 
To calculate the probability of finding no more than one error in the sample of 60 Ph.D.s, use
 


```python
poisson.cdf(1, mu=60 * 17 / 331)
```

Use the cumulative probability to decide how often a result of no more than one error would occur under this approximation.
 


## Exercise 1.4 Normal distribution


This exercise moves from counting errors to evaluating a monetary sample mean.
The audit task is to assess how unusual a sample mean of 1,012 is if the
population mean is assumed to be 1,030 and the standard deviation is estimated
at 115.26 for a sample of 200.
In Python we use `scipy.stats.norm.cdf()`. The argument `x` is the sampling
result that we evaluate, `loc` is the population mean, and `scale` is the
standard deviation of the mean.


```python
norm.cdf(x= 1012, loc= 1030, scale= 115.26 / sqrt(200))
```

Note that the syntax is more elaborate than what we have done so far. In
Python we can choose between passing argument values by position and passing
values specifically assigned to named arguments.
There are two advantages to using argument names and values. First, the command
is easier to interpret by a reviewer. Second, we can pass the arguments in any
order. The following commands are therefore equivalent:


```python
norm.cdf(1012, 1030, 115.26 / sqrt(200))
norm.cdf(x= 1012, loc= 1030, scale= 115.26 / sqrt(200))
norm.cdf(scale= 115.26 / sqrt(200), loc= 1030, x= 1012)
```

Because all three commands return the same probability, the learning point is about authoring reliable audit code: named arguments make the computation easier to review without changing the statistical result.
 


## Exercise 1.5 Student's $t$ distribution


This exercise evaluates a sample mean with a small sample size, where
uncertainty is represented with the $t$ distribution. The practical question is
how unusual the observed sample mean is under the assumed population mean.
To calculate the probability found of 14.7%, we use
`scipy.stats.t.cdf()` with arguments `x` for the boundary value and `df` for
the number of degrees of freedom.


```python
t_val = (1004 - 1030) / (73.8 / sqrt(10))
t.cdf(t_val, df=9)
```

 
The probability is 14.7%. Interpret this as the left-tail area for the observed sample mean. Notice that we used another shortcut notation: we added brackets to the assignment of the `t_val` variable, so that the assigned value is returned immediately.
 


## Exercise 1.6 $\chi^2$ (chi-squared) distribution


 
This exercise uses $\chi^2$ quantiles to build confidence bounds for a variance. The goal is to see how lower and upper distribution cutoffs become bounds on population variability.
The 95\% upper bound on a $\chi^2$ distributed variable with (10 - 1) degrees of freedom is
 


```python
chi2.ppf(.95, df=9)
```

 
Similarly, to calculate the upper bound for $\sigma^2$, we start with the lower bound of a $\chi^2$ distributed variable
 


```python
chi2.ppf(.05, df=9)
```

 
These two results are then used to calculate the upper and lower 90\% confidence bounds of the population variance.
Read the two cutoffs together. They define the range of variance values that remain plausible at the chosen confidence level.
 


## Exercise 1.7 $F$ distribution


This exercise calculates the decision boundary for an $F$ test. The practical
task is to find the critical value first, then use that boundary to decide what
would happen if a later test statistic falls above it.
Hypothesis testing is done with `scipy.stats.f.ppf()` or
`scipy.stats.f.isf()`. Here we use the upper-tail form because the rejection
region is in the right-hand tail of the $F$ distribution.


```python
f_crit = f.isf(0.05, dfn=25, dfd=23)
```

 
In the example above, the critical value of the $F$ statistic is the value shown in the output. Depending on whether the value of the test statistic is greater or smaller than this value, we then reject the null hypothesis or we do not. Before continuing, state the decision rule in words: if a later test statistic falls above this boundary, what conclusion follows?
 
The upper-tail probability at this critical value is calculated next.
 


```python
f.sf(f_crit, dfn=25, dfd=23)
```

The probability is the upper-tail area at the critical value. It should match the chosen significance level, which confirms that the critical value and test direction are aligned.
